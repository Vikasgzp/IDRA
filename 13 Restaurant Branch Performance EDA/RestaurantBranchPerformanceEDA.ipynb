```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_csv('/content/Day13_Restaurant_Branch_Performance_Dataset.csv')
```

## Initial Data Inspection

```python
# Display the first 5 rows of the DataFrame
print("First 5 rows of the dataset:")
display(df.head())
```

```python
# Get a concise summary of the DataFrame, including data types and non-null values
print("\nDataFrame Info:")
df.info()
```

```python
# Display descriptive statistics for numerical columns
print("\nDescriptive Statistics:")
display(df.describe())
```

## Distribution Analysis of Numerical Variables

```python
# Select numerical columns for distribution analysis
numerical_cols = ['Staff_Count', 'Marketing_Spend', 'Customers', 'Orders',
                  'Average_Bill', 'Revenue', 'Food_Cost', 'Operating_Cost',
                  'Profit', 'Avg_Delivery_Min', 'Customer_Rating',
                  'Temperature_C', 'Humidity_Percent']

# Plot histograms for numerical columns to visualize their distributions
plt.figure(figsize=(20, 15))
for i, col in enumerate(numerical_cols):
    plt.subplot(4, 4, i + 1) # Adjust subplot grid as needed
    sns.histplot(df[col], kde=True)
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')
plt.tight_layout()
plt.show()
```

## Performance Comparison Across Branches, Regions, and Store Types

```python
# Aggregate key metrics by Branch
print("\nPerformance by Branch:")
display(df.groupby('Branch')[['Revenue', 'Profit', 'Customers', 'Customer_Rating']].mean().sort_values(by='Profit', ascending=False))

# Aggregate key metrics by Region
print("\nPerformance by Region:")
display(df.groupby('Region')[['Revenue', 'Profit', 'Customers', 'Customer_Rating']].mean().sort_values(by='Profit', ascending=False))

# Aggregate key metrics by Store_Type
print("\nPerformance by Store Type:")
display(df.groupby('Store_Type')[['Revenue', 'Profit', 'Customers', 'Customer_Rating']].mean().sort_values(by='Profit', ascending=False))
```

## Correlation Analysis

```python
# Select numerical columns for correlation analysis
# We'll use the same list as before for consistency, excluding 'Record_ID'

numerical_cols_for_corr = ['Staff_Count', 'Marketing_Spend', 'Customers', 'Orders',
                           'Average_Bill', 'Revenue', 'Food_Cost', 'Operating_Cost',
                           'Profit', 'Avg_Delivery_Min', 'Customer_Rating',
                           'Temperature_C', 'Humidity_Percent']

# Calculate the correlation matrix
correlation_matrix = df[numerical_cols_for_corr].corr()

# Plot the correlation matrix as a heatmap
plt.figure(figsize=(16, 12))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Matrix of Numerical Variables')
plt.show()
```
