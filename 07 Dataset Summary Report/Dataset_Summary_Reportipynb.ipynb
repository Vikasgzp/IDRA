## Dataset Summary Report

This report aims to provide a comprehensive overview of the dataset's structure and characteristics using Python and the Pandas library.

```python
import pandas as pd

# Load the CSV dataset into a Pandas DataFrame
df = pd.read_csv('Day7_Student_Dataset.csv')

print("Dataset 'Day7_Student_Dataset.csv' loaded successfully.")
```

### 1. Inspecting the Dataset

Let's start by looking at the first and last few rows to get a sense of the data structure and content.

```python
print("First 5 rows:")
display(df.head())

print("\nLast 5 rows:")
display(df.tail())
```

### 2. Viewing Random Samples

Random samples can help in understanding the diversity of data entries.

```python
print("5 random rows from the dataset:")
display(df.sample(5))
```

### 3. Dataset Shape, Columns, and Index

Understanding the dimensions and identifiers of the dataset.

```python
print(f"Dataset shape (rows, columns): {df.shape}")
print(f"Number of rows: {df.shape[0]}")
print(f"Number of columns: {df.shape[1]}")

print("\nColumn names:")
display(df.columns.tolist())

print("\nDataset index:")
display(df.index)
```

### 4. Data Types and Non-Null Counts (using `info()`)

The `info()` method provides a concise summary of the DataFrame, including data types, non-null values, and memory usage. It's excellent for quickly identifying missing values and incorrect data types.

```python
df.info()
```

### 5. Descriptive Statistics (using `describe()`)

The `describe()` method generates descriptive statistics that summarize the central tendency, dispersion, and shape of a dataset's distribution, excluding `NaN` values.

```python
print("Descriptive statistics for numerical columns:")
display(df.describe())

print("\nDescriptive statistics for all columns (including categorical):")
display(df.describe(include='all'))
```

    *   `Marks`, `Attendance_Percentage`: Float (`float64`)
