# E-commerce Sales Analysis (Day 8)

```python
import pandas as pd

# Load the dataset into a Pandas DataFrame
df = pd.read_csv('/content/Day8_Ecommerce_Sales_Dataset.csv')

# Display the first 5 rows of the DataFrame
display(df.head())
```

## Data Inspection
Let's inspect the DataFrame to understand its structure, data types, and get a statistical summary.

```python
# Display information about the DataFrame, including data types and non-null values
df.info()

# Display descriptive statistics for numerical columns
display(df.describe())
```

## Data Cleaning and Preparation
Convert the 'Order_Date' column to datetime objects for easier time-based analysis.

```python
# Convert 'Order_Date' to datetime objects
df['Order_Date'] = pd.to_datetime(df['Order_Date'])

# Display the DataFrame info again to confirm the data type change
df.info()
```

## Sales Analysis by Category, City, and Product
Let's analyze total sales based on different categories, cities, and products.

```python
# Calculate total sales by Category
sales_by_category = df.groupby('Category')['Total_Sales'].sum().sort_values(ascending=False)
display(sales_by_category.to_frame())

# Calculate total sales by City
sales_by_city = df.groupby('City')['Total_Sales'].sum().sort_values(ascending=False)
display(sales_by_city.to_frame())

# Calculate total sales by Product
sales_by_product = df.groupby('Product')['Total_Sales'].sum().sort_values(ascending=False)
display(sales_by_product.to_frame())
```

## Sales Analysis by Payment Method

```python
# Calculate total sales by Payment Method
sales_by_payment = df.groupby('Payment_Method')['Total_Sales'].sum().sort_values(ascending=False)
display(sales_by_payment.to_frame())
```

## Key Sales Statistics
Let's calculate overall statistics for total sales, quantity sold, and number of orders.

```python
# Calculate total sales
total_sales = df['Total_Sales'].sum()

# Calculate average sales per order
average_sales = df['Total_Sales'].mean()

# Calculate highest sale amount
highest_sale = df['Total_Sales'].max()

# Calculate lowest sale amount
lowest_sale = df['Total_Sales'].min()

# Calculate total quantity sold
total_quantity_sold = df['Quantity'].sum()

# Calculate total number of orders (unique Order_IDs)
num_orders = df['Order_ID'].nunique()

print(f"Total Sales: ${total_sales:,.2f}")
print(f"Average Sales per Order: ${average_sales:,.2f}")
print(f"Highest Sale: ${highest_sale:,.2f}")
print(f"Lowest Sale: ${lowest_sale:,.2f}")
print(f"Total Quantity Sold: {total_quantity_sold}")
print(f"Number of Orders: {num_orders}")
```

## Identifying Top-Performing Products and Categories
Let's use sorting and filtering to find the top 5 products and categories by total sales.

```python
# Top 5 Products by Total Sales
top_5_products = df.groupby('Product')['Total_Sales'].sum().nlargest(5)
display(top_5_products.to_frame())

# Top 5 Categories by Total Sales
top_5_categories = df.groupby('Category')['Total_Sales'].sum().nlargest(5)
display(top_5_categories.to_frame())
```
