## Student Wellbeing Statistical Analysis & Probability

### 1. Data Loading and Initial Inspection

```python
import pandas as pd

# Load the dataset
df = pd.read_csv('Day16_Student_Wellbeing_Survey.csv')

# Display the first 5 rows of the DataFrame
print("First 5 rows of the dataset:")
display(df.head())
```

```python
# Display basic information about the DataFrame, including data types and non-null values
print("\nDataFrame Info:")
df.info()
```

```python
# Display descriptive statistics for numerical columns
print("\nDescriptive Statistics:")
display(df.describe())
```

### 2. Descriptive Statistics: Central Tendency (Mean, Median, Mode)

```python
variables_for_stats = [
    'Weekly_Study_Hours',
    'Average_Sleep_Hours',
    'Daily_Screen_Time_Hours',
    'Stress_Score',
    'Academic_Readiness_Score'
]

print("Mean, Median, and Mode for selected variables:")
for col in variables_for_stats:
    mean_val = df[col].mean()
    median_val = df[col].median()
    mode_val = df[col].mode().iloc[0] # .mode() can return multiple values, take the first

    print(f"\n--- {col} ---")
    print(f"Mean: {mean_val:.2f}")
    print(f"Median: {median_val:.2f}")
    print(f"Mode: {mode_val:.2f}")
```

### 3. Descriptive Statistics: Measures of Variability

```python
print("Measures of Variability for selected variables:")

variability_data = {}

for col in variables_for_stats:
    min_val = df[col].min()
    max_val = df[col].max()
    data_range = max_val - min_val
    variance_val = df[col].var()
    std_dev_val = df[col].std()
    q1_val = df[col].quantile(0.25)
    q3_val = df[col].quantile(0.75)
    iqr_val = q3_val - q1_val

    variability_data[col] = {'IQR': iqr_val}

    print(f"\n--- {col} ---")
    print(f"Range: {data_range:.2f}")
    print(f"Variance: {variance_val:.2f}")
    print(f"Standard Deviation: {std_dev_val:.2f}")
    print(f"Q1 (25th percentile): {q1_val:.2f}")
    print(f"Q3 (75th percentile): {q3_val:.2f}")
    print(f"IQR (Interquartile Range): {iqr_val:.2f}")

# Identify variable with the greatest variability (using IQR)
max_iqr_variable = max(variability_data, key=lambda k: variability_data[k]['IQR'])
print(f"\nVariable with the greatest variability (based on IQR): {max_iqr_variable} (IQR: {variability_data[max_iqr_variable]['IQR']:.2f})")
```

### 4. Outlier Detection using IQR Method

```python
import numpy as np

def detect_iqr_outliers(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)]
    non_outliers_df = df[(df[column] >= lower_bound) & (df[column] <= upper_bound)]

    print(f"\n--- Outlier Detection for {column} ---")
    print(f"Q1: {Q1:.2f}, Q3: {Q3:.2f}, IQR: {IQR:.2f}")
    print(f"Lower Bound: {lower_bound:.2f}, Upper Bound: {upper_bound:.2f}")
    print(f"Number of outliers detected: {len(outliers)}")
    if not outliers.empty:
        print(f"Outlier values (first 5):\n{outliers[column].head().to_string()}")
    return outliers, non_outliers_df

variables_for_outliers = [
    'Weekly_Study_Hours',
    'Daily_Screen_Time_Hours',
    'Commute_Time_Minutes',
    'Monthly_Discretionary_Spending'
]

outlier_dfs = {}
for var in variables_for_outliers:
    outliers, non_outliers_df = detect_iqr_outliers(df, var)
    outlier_dfs[var] = {'outliers': outliers, 'non_outliers_df': non_outliers_df}

# Compare mean and median before and after outlier removal for one variable (e.g., Weekly_Study_Hours)
print("\n--- Comparison of Mean and Median for Weekly_Study_Hours ---")
original_mean = df['Weekly_Study_Hours'].mean()
original_median = df['Weekly_Study_Hours'].median()
print(f"Before outlier removal - Mean: {original_mean:.2f}, Median: {original_median:.2f}")

weekly_study_non_outliers_df = outlier_dfs['Weekly_Study_Hours']['non_outliers_df']
cleaned_mean = weekly_study_non_outliers_df['Weekly_Study_Hours'].mean()
cleaned_median = weekly_study_non_outliers_df['Weekly_Study_Hours'].median()
print(f"After outlier removal - Mean: {cleaned_mean:.2f}, Median: {cleaned_median:.2f}")

print("Interpretation: Outliers can significantly affect the mean, pulling it towards extreme values. The median is more robust to outliers. In this case, removing outliers for Weekly_Study_Hours has slightly adjusted both the mean and median, indicating that the extreme values did have some influence on the central tendency.")
```

### 5. Probability Calculations

We define the following events:
- **Event A**: Student has a `Part_Time_Job` of "Yes"
- **Event B**: Student has `Stress_Score` ≥ 7
- **Event C**: Student has `Scholarship` = "Yes"
- **Event D**: Student `Exercise_Days_Per_Week` ≥ 3

```python
# Calculate P(A)
event_A = df[df['Part_Time_Job'] == 'Yes']
p_A = len(event_A) / len(df)
print(f"P(A) = P(Student has a Part_Time_Job = 'Yes'): {p_A:.4f}")

# Calculate P(B)
event_B = df[df['Stress_Score'] >= 7]
p_B = len(event_B) / len(df)
print(f"P(B) = P(Student has Stress_Score >= 7): {p_B:.4f}")

# Calculate P(C)
event_C = df[df['Scholarship'] == 'Yes']
p_C = len(event_C) / len(df)
print(f"P(C) = P(Student has Scholarship = 'Yes'): {p_C:.4f}")

# Calculate P(D)
event_D = df[df['Exercise_Days_Per_Week'] >= 3]
p_D = len(event_D) / len(df)
print(f"P(D) = P(Student exercises at least 3 days per week): {p_D:.4f}")
```

### 5.1 Joint, Conditional Probabilities and Event Relationships

```python
# Event A: Student has a Part_Time_Job of 'Yes'
# Event B: Student has Stress_Score >= 7

# Get the DataFrame for Event A and Event B from previous calculations
# event_A = df[df['Part_Time_Job'] == 'Yes'] # Already calculated in p_A
# event_B = df[df['Stress_Score'] >= 7] # Already calculated in p_B

# P(A and B): Intersection of Event A and Event B
event_A_and_B = df[(df['Part_Time_Job'] == 'Yes') & (df['Stress_Score'] >= 7)]
p_A_and_B = len(event_A_and_B) / len(df)
print(f"P(A and B) = P(Part_Time_Job='Yes' and Stress_Score>=7): {p_A_and_B:.4f}")

# P(A or B): Union of Event A and Event B
# Using the formula: P(A or B) = P(A) + P(B) - P(A and B)
p_A_or_B = p_A + p_B - p_A_and_B
print(f"P(A or B) = P(Part_Time_Job='Yes' or Stress_Score>=7): {p_A_or_B:.4f}")

# P(A|B): Probability of A given B
# P(A|B) = P(A and B) / P(B)
if p_B > 0:
    p_A_given_B = p_A_and_B / p_B
    print(f"P(A|B) = P(Part_Time_Job='Yes' | Stress_Score>=7): {p_A_given_B:.4f}")
else:
    print("P(B) is 0, so P(A|B) cannot be calculated.")

# P(B|A): Probability of B given A
# P(B|A) = P(A and B) / P(A)
if p_A > 0:
    p_B_given_A = p_A_and_B / p_A
    print(f"P(B|A) = P(Stress_Score>=7 | Part_Time_Job='Yes'): {p_B_given_A:.4f}")
else:
    print("P(A) is 0, so P(B|A) cannot be calculated.")

# Determine whether Year_of_Study = 1 and Year_of_Study = 4 are mutually exclusive events
event_Y1 = df[df['Year_of_Study'] == 1]
event_Y4 = df[df['Year_of_Study'] == 4]

# Check for intersection
intersection_Y1_Y4 = df[(df['Year_of_Study'] == 1) & (df['Year_of_Study'] == 4)]

print("\n--- Mutual Exclusivity Check ---")
if len(intersection_Y1_Y4) == 0:
    print("Events 'Year_of_Study = 1' and 'Year_of_Study = 4' are mutually exclusive because a student cannot be in both year 1 and year 4 simultaneously.")
else:
    print("Events 'Year_of_Study = 1' and 'Year_of_Study = 4' are NOT mutually exclusive.")

# Investigate independence of A and B
# A and B are independent if P(A and B) = P(A) * P(B)
product_p_A_p_B = p_A * p_B

print("\n--- Independence Check for Events A and B ---")
print(f"P(A and B): {p_A_and_B:.4f}")
print(f"P(A) * P(B): {product_p_A_p_B:.4f}")

tolerance = 1e-9 # A small tolerance for floating point comparisons
if abs(p_A_and_B - product_p_A_p_B) < tolerance:
    print("Conclusion: Events A and B appear to be independent because P(A and B) is approximately equal to P(A) * P(B).")
else:
    print("Conclusion: Events A and B do NOT appear to be independent because P(A and B) is not equal to P(A) * P(B).")
    print("This suggests that having a part-time job influences or is influenced by having a high stress score, or vice-versa.")
```

### 5.2 Bayes' Theorem

```python
# Event A: Student has a Part_Time_Job of 'Yes' (p_A already calculated)
# Event B: Student has Stress_Score >= 7 (p_B already calculated)
# P(A|B) and P(B|A) already calculated

# Calculate P(not A)
p_not_A = 1 - p_A
print(f"P(not A) = P(Student does NOT have a Part_Time_Job): {p_not_A:.4f}")

# Calculate P(B|not A): Probability of Stress_Score >= 7 given no Part_Time_Job
# Event (not A) = Student does NOT have a Part_Time_Job
event_not_A = df[df['Part_Time_Job'] == 'No']

# Event (B and not A) = Stress_Score >= 7 AND no Part_Time_Job
event_B_and_not_A = df[(df['Stress_Score'] >= 7) & (df['Part_Time_Job'] == 'No')]
p_B_and_not_A = len(event_B_and_not_A) / len(df)

if len(event_not_A) > 0:
    p_B_given_not_A = p_B_and_not_A / p_not_A
    print(f"P(B|not A) = P(Stress_Score>=7 | no Part_Time_Job): {p_B_given_not_A:.4f}")
else:
    print("P(not A) is 0, so P(B|not A) cannot be calculated.")
    p_B_given_not_A = 0 # Assign 0 if cannot calculate to avoid errors in Bayes' theorem if used

# Verify P(B) using Law of Total Probability: P(B) = P(B|A)P(A) + P(B|not A)P(not A)
# p_B_total_prob = (p_B_given_A * p_A) + (p_B_given_not_A * p_not_A)
# print(f"P(B) via Law of Total Probability: {p_B_total_prob:.4f}")
# print(f"Original P(B): {p_B:.4f}")

# Apply Bayes' Theorem to calculate P(A|B):
# P(A|B) = [P(B|A) * P(A)] / P(B)
# We know P(B) = P(B|A)P(A) + P(B|not A)P(not A)

# Calculate denominator P(B)
denominator_p_B = (p_B_given_A * p_A) + (p_B_given_not_A * p_not_A)

print(f"\n--- Bayes' Theorem Calculation for P(A|B) ---")
print(f"Formula: P(A|B) = [P(B|A) * P(A)] / [P(B|A)P(A) + P(B|not A)P(not A)]")
print(f"P(B|A): {p_B_given_A:.4f}")
print(f"P(A): {p_A:.4f}")
print(f"P(B|not A): {p_B_given_not_A:.4f}")
print(f"P(not A): {p_not_A:.4f}")

if denominator_p_B > 0:
    p_A_given_B_bayes = (p_B_given_A * p_A) / denominator_p_B
    print(f"P(A|B) using Bayes' Theorem: {p_A_given_B_bayes:.4f}")
else:
    print("Denominator P(B) is 0, so P(A|B) using Bayes' Theorem cannot be calculated.")
    p_A_given_B_bayes = np.nan # Assign NaN if cannot calculate

# Verify with direct calculation of P(A|B)
print(f"P(A|B) from direct calculation: {p_A_given_B:.4f}")

# Compare and explain
if abs(p_A_given_B_bayes - p_A_given_B) < tolerance:
    print("Interpretation: The P(A|B) calculated using Bayes' Theorem agrees with the direct conditional probability calculation, confirming the consistency of our probability estimations.")
else:
    print("Interpretation: The P(A|B) calculated using Bayes' Theorem does NOT agree with the direct conditional probability calculation. There might be an error in the calculations.")
```

### 6. Normal Distribution Analysis of Academic Readiness Score

```python
# Calculate mean and standard deviation for Academic_Readiness_Score
academic_readiness_mean = df['Academic_Readiness_Score'].mean()
academic_readiness_std = df['Academic_Readiness_Score'].std()

print(f"Mean Academic Readiness Score: {academic_readiness_mean:.2f}")
print(f"Standard Deviation Academic Readiness Score: {academic_readiness_std:.2f}")

# Find the Z-score for the highest score
highest_score = df['Academic_Readiness_Score'].max()
highest_score_z = (highest_score - academic_readiness_mean) / academic_readiness_std
print(f"\nHighest Academic Readiness Score: {highest_score:.2f}")
print(f"Z-score for highest score: {highest_score_z:.2f}")
print("Interpretation: A positive Z-score indicates the score is above the mean. A Z-score greater than 2 or 3 suggests an unusually high score compared to the average.")

# Find the Z-score for the lowest score
lowest_score = df['Academic_Readiness_Score'].min()
lowest_score_z = (lowest_score - academic_readiness_mean) / academic_readiness_std
print(f"\nLowest Academic Readiness Score: {lowest_score:.2f}")
print(f"Z-score for lowest score: {lowest_score_z:.2f}")
print("Interpretation: A negative Z-score indicates the score is below the mean. A Z-score less than -2 or -3 suggests an unusually low score compared to the average.")

# Use the 68-95-99.7 empirical rule
print("\n--- 68-95-99.7 Empirical Rule for Academic Readiness Score ---")

# Within 1 standard deviation
lower_1_std = academic_readiness_mean - academic_readiness_std
upper_1_std = academic_readiness_mean + academic_readiness_std
print(f"Approximately 68% of students are expected to have an Academic Readiness Score between {lower_1_std:.2f} and {upper_1_std:.2f}.")

# Within 2 standard deviations
lower_2_std = academic_readiness_mean - (2 * academic_readiness_std)
upper_2_std = academic_readiness_mean + (2 * academic_readiness_std)
print(f"Approximately 95% of students are expected to have an Academic Readiness Score between {lower_2_std:.2f} and {upper_2_std:.2f}.")

# Within 3 standard deviations
lower_3_std = academic_readiness_mean - (3 * academic_readiness_std)
upper_3_std = academic_readiness_mean + (3 * academic_readiness_std)
print(f"Approximately 99.7% of students are expected to have an Academic Readiness Score between {lower_3_std:.2f} and {upper_3_std:.2f}.")
```

### 7. Conclusion: 5 Meaningful Statistical Observations

Based on the comprehensive statistical analysis of the 'Student Wellbeing Survey' dataset, here are 5 meaningful statistical observations:

1.  **Academic Readiness Score Variability**: The `Academic_Readiness_Score` exhibits the highest variability among the assessed quantitative metrics (IQR: 12.83). This indicates a wide range of academic preparedness within the student population, suggesting that interventions or support programs might need to be tailored to diverse academic levels.

2.  **Impact of Outliers on Weekly Study Hours**: The comparison of mean and median for `Weekly_Study_Hours` before (Mean: 15.69, Median: 15.40) and after (Mean: 15.60, Median: 15.35) outlier removal, though showing slight changes, confirms that outliers can subtly influence central tendency measures. While the median remained relatively stable, the mean showed a small adjustment, indicating that extreme study hours exist but might not drastically skew the overall central tendency for this variable.

3.  **Stress and Part-Time Jobs are Not Independent**: Events A (having a part-time job) and B (having a stress score >= 7) were found to be *not* independent (P(A and B) = 0.0517 vs. P(A) * P(B) = 0.0190). This significant difference suggests a relationship between having a part-time job and experiencing higher stress levels. Students with part-time jobs are more likely to experience higher stress (P(B|A) = 0.2039) compared to the overall student population (P(B) = 0.0750).

4.  **High Stress Among Part-Time Workers**: The conditional probability P(A|B) (probability of having a part-time job given a high stress score) was calculated as 0.6889. This high probability suggests that a substantial majority of students experiencing high stress also hold part-time jobs, reinforcing the previous observation about the strong association between these two factors. This implies that managing work-life balance could be a critical factor in student well-being.

5.  **Normal Distribution of Academic Readiness**: The `Academic_Readiness_Score` appears to follow a reasonably normal distribution, as indicated by the Z-scores and the empirical rule. Approximately 68% of students score between 62.08 and 81.47, and 95% score between 52.38 and 91.16. The highest score (98.00) has a Z-score of 2.71, and the lowest score (41.90) has a Z-score of -3.08, suggesting that while extreme scores exist, most students fall within 2-3 standard deviations of the mean. This allows for the use of normal distribution assumptions for further analysis or predictions related to academic readiness.
