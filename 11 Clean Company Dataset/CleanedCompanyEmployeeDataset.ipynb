```python
import pandas as pd

# Load the dataset
file_path = '/content/Day11_Messy_Company_Employee_Dataset.csv'
df = pd.read_csv(file_path)

print('Dataset loaded successfully. Here are the first 5 rows:')
display(df.head())
```

```python
print('\nDataset Information:')
df.info()
```

```python
print('\nDescriptive Statistics:')
display(df.describe(include='all'))
```

```python
# Drop duplicate Employee_ID entries, keeping the first occurrence
df.drop_duplicates(subset='Employee_ID', keep='first', inplace=True)
print('Duplicate Employee_IDs removed. New number of rows:', len(df))
```

```python
# Standardize 'Work_Mode' column
df['Work_Mode'] = df['Work_Mode'].replace({'In-office': 'Office', 'Work From Home': 'Remote'})
print('\nWork_Mode values after standardization:')
print(df['Work_Mode'].value_counts(dropna=False))
```

```python
# Further standardize 'Work_Mode' by converting to consistent case (e.g., capitalize first letter)
df['Work_Mode'] = df['Work_Mode'].str.capitalize()
print('\nWork_Mode values after complete standardization:')
print(df['Work_Mode'].value_counts(dropna=False))
```

```python
# Standardize 'Department' column
# Check unique values first to identify inconsistencies
print('\nUnique values for Department before standardization:')
print(df['Department'].unique())

# Assuming 'Data Science' and 'Research' might be intended to be distinct, but let's check further if any overlaps.
# For now, let's just make sure they are consistently cased if needed, or replace common misspellings.
# As per initial observation, 'Data Science' and 'Research' are distinct. Let's make sure all are capitalized.
df['Department'] = df['Department'].str.capitalize()
print('\nDepartment values after standardization (capitalize):')
print(df['Department'].value_counts(dropna=False))
```

```python
# Quantify missing values again after initial cleaning steps
print('\nMissing values after removing duplicates and standardizing Work_Mode/Department:')
display(df.isnull().sum())
```

```python
# Further standardize 'Department' to handle 'Customer Success' correctly
df['Department'] = df['Department'].replace('Customer success', 'Customer Success')
print('\nDepartment values after further standardization:')
print(df['Department'].value_counts(dropna=False))
```

```python
# Impute missing values

# Categorical columns: impute with mode
for col in ['Department', 'Gender', 'City', 'Work_Mode']:
    mode_val = df[col].mode()[0]
    df[col].fillna(mode_val, inplace=True)
    print(f'Missing values in {col} imputed with mode: {mode_val}')

# Numerical columns: impute with median
for col in ['Age', 'Annual_Salary', 'Experience_Years', 'Performance_Score']:
    median_val = df[col].median()
    df[col].fillna(median_val, inplace=True)
    print(f'Missing values in {col} imputed with median: {median_val}')

print('\nMissing values after imputation:')
display(df.isnull().sum())
```

```python
# Convert 'Joining_Date' to datetime objects and verify data types
df['Joining_Date'] = pd.to_datetime(df['Joining_Date'], errors='coerce')

print('\nData types after conversion:')
df.info()
```

```python
# Final verification of the dataset after cleaning
print('\nFinal check of missing values:')
display(df.isnull().sum())

print('\nFinal check of data types:')
df.info()

print('\nFirst 5 rows of the cleaned dataset:')
display(df.head())
```

```python
# Export the cleaned dataset to a new CSV file
output_file_path = '/content/Cleaned_Company_Employee_Dataset.csv'
df.to_csv(output_file_path, index=False)
print(f'\nCleaned dataset exported to {output_file_path}')
```
