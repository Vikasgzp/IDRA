## Flight Operations Data Analysis

This notebook will guide you through a complete data analysis workflow on the provided Flight Operations dataset using Pandas. We will cover loading, inspecting, cleaning, transforming, and analyzing the data to identify patterns and key findings.

### 1. Load and Inspect the Dataset

```python
import pandas as pd

# Load the dataset
file_path = 'Day10_Flight_Operations_Dataset.csv'
try:
    df = pd.read_csv(file_path)
    print(f"Dataset '{file_path}' loaded successfully.")
except FileNotFoundError:
    print(f"Error: The file '{file_path}' was not found. Please ensure it's uploaded correctly.")
    df = None
```

```python
if df is not None:
    # Display the first 5 rows of the dataset
    print("\nFirst 5 rows of the dataset:")
    display(df.head())

    # Display basic information about the dataset (columns, non-null counts, dtypes)
    print("\nDataset Information:")
    df.info()

    # Display descriptive statistics
    print("\nDescriptive Statistics:")
    display(df.describe(include='all'))
```

### 3. Data Transformation and Feature Engineering

```python
if df is not None:
    print("\n--- Data Transformation ---")

    # 1. Create 'Occupancy_Rate' feature
    df['Occupancy_Rate'] = (df['Passengers'] / df['Seat_Capacity']) * 100
    print("Created 'Occupancy_Rate' column.")

    # 2. Extract 'Day_of_Week' from 'Flight_Date'
    df['Day_of_Week'] = df['Flight_Date'].dt.day_name()
    df['Day_of_Week'] = df['Day_of_Week'].astype('category') # Convert to category for efficiency
    print("Created 'Day_of_Week' column.")

    # Display the first few rows with new features
    print("\nFirst 5 rows with new features:")
    display(df.head())
```

### 4. Exploratory Data Analysis (EDA) - Overall Metrics

```python
if df is not None:
    print("\n--- Overall Flight Operations Metrics ---")

    # Calculate overall average delay
    avg_delay = df['Delay_Minutes'].mean()
    print(f"Average Flight Delay: {avg_delay:.2f} minutes")

    # Calculate overall average ticket price
    avg_ticket_price = df['Average_Ticket_Price'].mean()
    print(f"Average Ticket Price: ${avg_ticket_price:.2f}")

    # Calculate overall average occupancy rate
    avg_occupancy_rate = df['Occupancy_Rate'].mean()
    print(f"Average Occupancy Rate: {avg_occupancy_rate:.2f}%")

    # Calculate overall average passenger satisfaction
    avg_satisfaction = df['Passenger_Satisfaction'].mean()
    print(f"Average Passenger Satisfaction (1-5 scale): {avg_satisfaction:.2f}")

    # Calculate the percentage of delayed flights
    total_flights = df.shape[0]
    delayed_flights = df[df['Flight_Status'] == 'Delayed'].shape[0]
    percentage_delayed = (delayed_flights / total_flights) * 100
    print(f"Percentage of Delayed Flights: {percentage_delayed:.2f}%")
```

### 5. Analyzing Trends by Categories (Grouping and Aggregation)

#### 5.1. Performance by Airline

```python
if df is not None:
    print("\n--- Airline Performance Analysis ---")
    airline_performance = df.groupby('Airline').agg(
        Average_Delay=('Delay_Minutes', 'mean'),
        Average_Ticket_Price=('Average_Ticket_Price', 'mean'),
        Average_Passenger_Satisfaction=('Passenger_Satisfaction', 'mean'),
        Total_Flights=('Flight_ID', 'count')
    ).sort_values(by='Total_Flights', ascending=False)

    print("\nAirline Performance (sorted by Total Flights):")
    display(airline_performance)
```

#### 5.3. Impact of Weather Conditions

```python
if df is not None:
    print("\n--- Weather Impact Analysis ---")
    weather_impact = df.groupby('Weather').agg(
        Average_Delay=('Delay_Minutes', 'mean'),
        Average_Satisfaction=('Passenger_Satisfaction', 'mean'),
        Total_Flights=('Flight_ID', 'count')
    ).sort_values(by='Average_Delay', ascending=False)

    print("\nAnalysis by Weather Condition:")
    display(weather_impact)
```

### 6. Key Findings and Observations

Based on the analysis, here are some key observations:

1.  **Overall Operations**: The average flight delay across all operations is **18.78 minutes**, with **32.78%** of all flights experiencing a delay, indicating that delays are a significant operational challenge. The average passenger satisfaction is moderate at **3.49 out of 5**, and the average occupancy rate stands at **62.66%**.

2.  **Impact of Flight Status on Satisfaction**: Unsurprisingly, **Delayed flights have the lowest average passenger satisfaction (3.27)** compared to On-Time flights (3.59) and even Cancelled flights (3.86). This suggests that effective delay management and communication are crucial for passenger experience. Interestingly, cancelled flights have a higher satisfaction, possibly due to prompt refunds or alternative arrangements.

3.  **Travel Class Dynamics**: **Business Class has significantly higher average ticket prices ($16,780)** than Premium Economy ($8,736) and Economy ($4,550). While Business Class has a similar occupancy rate to Premium Economy (around 67%), its passenger satisfaction (3.27) is notably lower than Premium Economy (3.69) and Economy (3.46). This indicates that higher-paying Business Class passengers may have higher expectations that are not always met.

4.  **Airline Performance Variation**: Among airlines, **Air India Express and Air India exhibit the highest average passenger satisfaction (3.67 and 3.75 respectively)**, despite Air India having a relatively higher average delay (20.58 minutes). IndiGo, while having a high number of flights, shows a higher average delay (25.03 minutes) and slightly lower satisfaction (3.34). Vistara has the most flights but an average delay of 17.89 minutes and satisfaction of 3.38.

5.  **Weather's Role in Delays**: **Rainy weather leads to the highest average delays (21.42 minutes)**, closely followed by Stormy conditions (19.60 minutes). Cloudy and Clear weather conditions have similar average delays (around 18 minutes), while Fog results in the lowest average delay (15.49 minutes). Passenger satisfaction also tends to be lower during adverse weather, with Stormy weather having the lowest satisfaction (3.17).

6.  **Occupancy and Pricing**: While Economy class has the lowest average ticket price, it also has the lowest average occupancy rate (60.98%) compared to Business and Premium Economy (both around 67%). This suggests potential opportunities to optimize pricing or capacity in Economy class to boost occupancy.

7.  **Key Determinants of Satisfaction**: Delay minutes and travel class appear to be strong determinants of passenger satisfaction. Minimizing delays, especially during adverse weather, and addressing the specific expectations of Business Class passengers could significantly improve overall satisfaction.

#### 5.2. Impact of Flight Status and Travel Class

```python
if df is not None:
    print("\n--- Flight Status Analysis ---")
    flight_status_analysis = df.groupby('Flight_Status').agg(
        Average_Delay=('Delay_Minutes', 'mean'),
        Average_Satisfaction=('Passenger_Satisfaction', 'mean'),
        Total_Flights=('Flight_ID', 'count')
    )
    print("\nAnalysis by Flight Status:")
    display(flight_status_analysis)

    print("\n--- Travel Class Analysis ---")
    travel_class_analysis = df.groupby('Travel_Class').agg(
        Average_Ticket_Price=('Average_Ticket_Price', 'mean'),
        Average_Occupancy_Rate=('Occupancy_Rate', 'mean'),
        Average_Satisfaction=('Passenger_Satisfaction', 'mean'),
        Total_Flights=('Flight_ID', 'count')
    ).sort_values(by='Average_Ticket_Price', ascending=False)

    print("\nAnalysis by Travel Class:")
    display(travel_class_analysis)
```

### 2. Data Cleaning and Type Conversion

```python
if df is not None:
    print("\n--- Data Cleaning and Type Conversion ---")

    # 1. Handle Missing Values in 'Delay_Minutes'
    # Impute missing 'Delay_Minutes' with the median value
    median_delay = df['Delay_Minutes'].median()
    df['Delay_Minutes'].fillna(median_delay, inplace=True)
    print(f"Missing values in 'Delay_Minutes' imputed with median: {median_delay}.")

    # 2. Convert 'Flight_Date' to datetime objects
    df['Flight_Date'] = pd.to_datetime(df['Flight_Date'])
    print("Converted 'Flight_Date' to datetime objects.")

    # 3. Convert object columns to 'category' type for memory efficiency and better analysis
    categorical_cols = [
        'Airline', 'Origin', 'Destination', 'Aircraft',
        'Travel_Class', 'Flight_Status', 'Weather',
        'Booking_Channel', 'Meal_Preference'
    ]
    for col in categorical_cols:
        df[col] = df[col].astype('category')
    print("Converted several object columns to 'category' type.")

    # Verify changes
    print("\nUpdated Dataset Information:")
    df.info()
    print("\nNumber of missing values after cleaning:")
    print(df.isnull().sum())
```
