### Load Data

I'll load the three provided CSV files (`Day9_Orders.csv`, `Day9_Customers.csv`, and `Day9_Products.csv`) into pandas DataFrames. These files are essential for creating the processed e-commerce dataset.

```python
import pandas as pd

# Load the datasets
orders_df = pd.read_csv('Day9_Orders.csv')
customers_df = pd.read_csv('Day9_Customers.csv')
products_df = pd.read_csv('Day9_Products.csv')

print('Orders DataFrame head:')
display(orders_df.head())

print('\nCustomers DataFrame head:')
display(customers_df.head())

print('\nProducts DataFrame head:')
display(products_df.head())
```

### Combine DataFrames using `merge()`

I will now merge the `orders_df`, `customers_df`, and `products_df` to create a unified dataset. This will involve two `merge()` operations:
1. Merge `orders_df` with `customers_df` on `Customer_ID`.
2. Merge the result with `products_df` on `Product_ID`.

```python
# Merge orders and customers data
merged_df = pd.merge(orders_df, customers_df, on='Customer_ID', how='left')

# Merge the result with products data
merged_df = pd.merge(merged_df, products_df, on='Product_ID', how='left')

print('Merged DataFrame head:')
display(merged_df.head())

print('\nMerged DataFrame info:')
display(merged_df.info())
```

### Demonstrate `concat()`

To fulfill the assignment's requirement of demonstrating `concat()`, I will create two small, temporary DataFrames and then concatenate them. This illustrates how `pd.concat()` can be used to stack or combine DataFrames.

```python
# Create two sample DataFrames for concat demonstration
df1 = pd.DataFrame({'A': ['A0', 'A1'], 'B': ['B0', 'B1']})
df2 = pd.DataFrame({'A': ['A2', 'A3'], 'B': ['B2', 'B3']})

# Concatenate the two DataFrames
concatenated_df = pd.concat([df1, df2], ignore_index=True)

print('DataFrame 1:')
display(df1)

print('\nDataFrame 2:')
display(df2)

print('\nConcatenated DataFrame:')
display(concatenated_df)
```

### DateTime Operations and Feature Engineering

Now, I will perform DateTime operations on the `Order_Date` column in the `merged_df`. This involves:
1. Converting the `Order_Date` column to datetime objects.
2. Extracting the month, day, and day of the week from the `Order_Date`.

```python
# Convert 'Order_Date' to datetime objects
merged_df['Order_Date'] = pd.to_datetime(merged_df['Order_Date'])

# Extract month, day, and day of the week
merged_df['Order_Month'] = merged_df['Order_Date'].dt.month
merged_df['Order_Day'] = merged_df['Order_Date'].dt.day
merged_df['Order_DayOfWeek'] = merged_df['Order_Date'].dt.day_name()

print('DataFrame with new date features:')
display(merged_df[['Order_Date', 'Order_Month', 'Order_Day', 'Order_DayOfWeek']].head())
```

### Create or Transform Useful Columns using `apply()`

I will now use the `apply()` method to create new, useful columns:
1. **`Total_Price`**: Calculate the total price for each item by multiplying `Quantity` and `Unit_Price`.
2. **`Order_Value`**: Apply a discount based on `Membership_Type` to demonstrate conditional logic with `apply()`:
    - Premium members get a 10% discount.
    - Regular members get a 5% discount.
    - New members get no discount.

```python
# Calculate Total_Price for each item
merged_df['Total_Price'] = merged_df.apply(lambda row: row['Quantity'] * row['Unit_Price'], axis=1)

# Define a function to calculate Order_Value with discount based on Membership_Type
def calculate_order_value(row):
    total_price = row['Total_Price']
    membership = row['Membership_Type']

    if membership == 'Premium':
        return total_price * 0.90  # 10% discount
    elif membership == 'Regular':
        return total_price * 0.95  # 5% discount
    else: # New or any other type
        return total_price

# Apply the function to create Order_Value
merged_df['Order_Value'] = merged_df.apply(calculate_order_value, axis=1)

print('DataFrame with new calculated columns:')
display(merged_df[['Quantity', 'Unit_Price', 'Total_Price', 'Membership_Type', 'Order_Value']].head())
```

### Organize and Export Final Dataset

Finally, I will organize the `merged_df` into a clean and meaningful DataFrame by selecting and reordering the columns. After organizing, I will export the final processed dataset to a CSV file as requested.

```python
# Select and reorder columns for the final processed dataset
final_df = merged_df[[
    'Order_ID', 'Order_Date', 'Order_Month', 'Order_Day', 'Order_DayOfWeek',
    'Customer_ID', 'Customer_Name', 'City', 'Region', 'Membership_Type',
    'Product_ID', 'Product_Name', 'Category', 'Brand', 'Unit_Price',
    'Quantity', 'Total_Price', 'Order_Value', 'Payment_Method', 'Order_Status'
]]

print('Final Processed DataFrame head:')
display(final_df.head())

# Export the final DataFrame to a CSV file
final_df.to_csv('processed_ecommerce_dataset.csv', index=False)

print('\nFinal processed dataset exported to processed_ecommerce_dataset.csv')

```
